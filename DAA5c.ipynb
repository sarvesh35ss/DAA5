```python
import random

def quickselect(arr, k):
    # Convert k-th smallest to 0-based index
    k = k - 1

    def select(left, right):
        # Base case
        if left == right:
            return arr[left]

        # Choose a random pivot
        pivot_index = random.randint(left, right)
        pivot = arr[pivot_index]

        # Move pivot to the end
        arr[pivot_index], arr[right] = arr[right], arr[pivot_index]

        # Partition
        store_index = left

        for i in range(left, right):
            if arr[i] < pivot:
                arr[i], arr[store_index] = arr[store_index], arr[i]
                store_index += 1

        # Put pivot in its correct position
        arr[store_index], arr[right] = arr[right], arr[store_index]

        # Divide and Conquer
        if k == store_index:
            return arr[store_index]

        elif k < store_index:
            return select(left, store_index - 1)

        else:
            return select(store_index + 1, right)

    return select(0, len(arr) - 1)


# ------------------------------------------------
# Test with 1000 random integers
# ------------------------------------------------

n = 1000

arr = [random.randint(1, 10000) for _ in range(n)]

print("Array size:", n)

# k = 1 → minimum
result_min = quickselect(arr.copy(), 1)
print("1st smallest (minimum):", result_min)

# k = n → maximum
result_max = quickselect(arr.copy(), n)
print("1000th smallest (maximum):", result_max)

# k = n//2 → median position
k = n // 2
result_median = quickselect(arr.copy(), k)
print(f"{k}th smallest (median position):", result_median)
```
