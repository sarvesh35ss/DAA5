```python
import random

# --------------------------------------------------
# Divide and Conquer approach
# --------------------------------------------------

def divide_conquer_min_max(arr, low, high):
    # Only one element
    if low == high:
        return arr[low], arr[low], 0

    # Two elements
    if high == low + 1:
        comparisons = 1

        if arr[low] < arr[high]:
            return arr[low], arr[high], comparisons
        else:
            return arr[high], arr[low], comparisons

    # Divide
    mid = (low + high) // 2

    min1, max1, comp1 = divide_conquer_min_max(arr, low, mid)
    min2, max2, comp2 = divide_conquer_min_max(arr, mid + 1, high)

    # Combine
    comparisons = comp1 + comp2

    comparisons += 1
    minimum = min1 if min1 < min2 else min2

    comparisons += 1
    maximum = max1 if max1 > max2 else max2

    return minimum, maximum, comparisons


# --------------------------------------------------
# Naive approach
# --------------------------------------------------

def naive_min_max(arr):
    minimum = arr[0]
    maximum = arr[0]
    comparisons = 0

    for i in range(1, len(arr)):
        comparisons += 1
        if arr[i] < minimum:
            minimum = arr[i]

        comparisons += 1
        if arr[i] > maximum:
            maximum = arr[i]

    return minimum, maximum, comparisons


# --------------------------------------------------
# Testing
# --------------------------------------------------

sizes = [10, 100, 1000, 10000]

print("Size\tD&C Comparisons\tNaive Comparisons")
print("-" * 50)

for n in sizes:

    arr = [random.randint(1, 100000) for _ in range(n)]

    # Divide and Conquer
    min_dc, max_dc, comp_dc = divide_conquer_min_max(
        arr, 0, n - 1
    )

    # Naive
    min_naive, max_naive, comp_naive = naive_min_max(arr)

    print(f"{n}\t{comp_dc}\t\t{comp_naive}")
```
