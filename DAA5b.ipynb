```python
def min_max_second(arr, low, high):
    # Base case: one element
    if low == high:
        return arr[low], float('inf'), arr[low], float('-inf'), 0

    # Base case: two elements
    if high == low + 1:
        a = arr[low]
        b = arr[high]

        comparisons = 1

        if a < b:
            return a, b, b, a, comparisons
        else:
            return b, a, a, b, comparisons

    # Divide
    mid = (low + high) // 2

    min1, second_min1, max1, second_max1, c1 = \
        min_max_second(arr, low, mid)

    min2, second_min2, max2, second_max2, c2 = \
        min_max_second(arr, mid + 1, high)

    comparisons = c1 + c2

    # ------------------------------------------------
    # Find minimum and second minimum
    # ------------------------------------------------

    comparisons += 1

    if min1 < min2:
        minimum = min1
        second_minimum = min(second_min1, min2)
        comparisons += 1
    else:
        minimum = min2
        second_minimum = min(second_min2, min1)
        comparisons += 1

    # ------------------------------------------------
    # Find maximum and second maximum
    # ------------------------------------------------

    comparisons += 1

    if max1 > max2:
        maximum = max1
        second_maximum = max(second_max1, max2)
        comparisons += 1
    else:
        maximum = max2
        second_maximum = max(second_max2, max1)
        comparisons += 1

    return minimum, second_minimum, maximum, second_maximum, comparisons


# Example
arr = [12, 5, 8, 20, 3, 15, 7, 10]

minimum, second_minimum, maximum, second_maximum, comparisons = \
    min_max_second(arr, 0, len(arr) - 1)

print("Array:", arr)
print("Minimum:", minimum)
print("Second Minimum:", second_minimum)
print("Maximum:", maximum)
print("Second Maximum:", second_maximum)
print("Comparisons:", comparisons)
```
